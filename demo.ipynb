# Anchor Lattice Filter — Module A demo

Frozen spec **v1.0.0** (2026-09-14). This notebook is the live demo.

It keep/drops a small batch with the public filter: entropy gate, opposition balance, duplicate gate, score gate. Default backend is the numpy character 3-gram hash, so it runs without MiniLM.

This is **not** the Anchor Memory diary, **not** REFLECTION GATE, and **not** the icosahedral phonon notebooks. It does not measure hallucination rate, deliberation speed, or data efficiency. Those are not in this run. Held-out counts are in `RESULTS.md`.

A line under 12 characters drops as `short` before the other gates. That includes a tweet-length fragment.

```python
import json
from anchor_lattice import filter_batch

batch = json.load(open("examples/batch.json", encoding="utf-8"))
texts = [row["text"] for row in batch]
poles = [row.get("pole", "none") for row in batch]
result = filter_batch(texts, poles=poles, backend="hash")
print("spec", result["spec_version"], "backend", result["backend"])
print("kept", result["kept"], "dropped", result["dropped"], "batch_balance", result["batch_balance"])
print()
for item in result["items"]:
    flag = "KEEP" if item["keep"] else "DROP"
    print(f"{flag:4} {item['reason']:13} H={item['H_hat']:.3f} B={item['B']:.3f} U={item['U']:.3f} W={item['W']:.3f}  {item['text'][:72]}")
```

Expected on this batch: the duplicate of the first line drops as `duplicate`, and `too short` drops as `short`. Everything else is a score, not a slogan.
